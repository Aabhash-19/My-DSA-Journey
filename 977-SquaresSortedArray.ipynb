```python
# Brute Force
class Solution:
    def sortedSquares(self, nums: List[int]) -> List[int]:
        for i in range(0,len(nums)):
            nums[i]=nums[i]**2
        nums.sort()
        return nums   
```

```python
# Optimized
class Solution:
    def sortedSquares(self, nums: List[int]) -> List[int]:
        arr_pos=[]
        arr_neg=[]
        
        for i in range(0,len(nums)):
            if nums[i]>=0:
                arr_pos.append(nums[i])
            else:
                arr_neg.append(nums[i])
        
        if len(arr_neg)==0:
            for i in range(0,len(nums)):
                nums[i]=nums[i]**2
            
        elif len(arr_pos)==0:
            for i in range(0,len(nums)):
                nums[i]=nums[i]**2
            nums.reverse()
            
        for i in range(0,len(arr_pos)):
            arr_pos[i]=arr_pos[i]**2
        
        for j in range(0,len(arr_neg)):
            arr_neg[j]=arr_neg[j]**2
        arr_neg.reverse()

        res=[]
        n,m=len(arr_pos),len(arr_neg)
        i=j=0

        while(i<n and j<m):
            if arr_pos[i]<arr_neg[j]:
                res.append(arr_pos[i])
                i=i+1
            else:
                res.append(arr_neg[j])
                j=j+1

        while(i<n):
            res.append(arr_pos[i])
            i=i+1

        while(j<m):
            res.append(arr_neg[j])
            j=j+1

        return res
```
