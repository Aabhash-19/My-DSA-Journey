```python
class Solution:
    def removeDuplicates(self, nums: List[int]) -> int:
       low=0
       unique=1
       high=1

       while(high<len(nums)):
        if(nums[high]==nums[high-1]):
            high=high+1
            continue
        else:
            nums[low+1]=nums[high]
            unique=unique+1
            high=high+1
            low=low+1
       return unique
```
