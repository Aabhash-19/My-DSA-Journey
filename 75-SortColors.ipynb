```python
# Brute Force

class Solution:
    def sortColors(self, nums: List[int]) -> None:
        arr_0=[]
        arr_1=[]
        arr_2=[]

        for i in range(0,len(nums)):
            if nums[i]==0:
                arr_0.append(nums[i])

            elif nums[i]==1:
                arr_1.append(nums[i])    

            elif nums[i]==2:
                arr_2.append(nums[i])    

        nums[:]=arr_0+arr_1+arr_2
```

```python
# Optimized

class Solution:
    def sortColors(self, nums: List[int]) -> None:
        zero=0
        one=0
        two=0

        for i in range(0,len(nums)):
            if nums[i]==0:
                zero=zero+1
            
            elif nums[i]==1:
                one=one+1

            elif nums[i]==2:
                two=two+1

        nums[:]=[0]*zero+[1]*one+[2]*two
```

```python
# Dutch National Flag Algo

class Solution:
    def sortColors(self, nums: List[int]) -> None:
        low=0
        mid=0
        high=len(nums)-1

        while mid<=high:
            if nums[mid]==0:
                nums[low],nums[mid]=nums[mid],nums[low]
                low=low+1
                mid=mid+1

            elif nums[mid]==1:
                mid=mid+1

            elif nums[mid]==2:
                nums[mid],nums[high]=nums[high],nums[mid]
                high=high-1
```
