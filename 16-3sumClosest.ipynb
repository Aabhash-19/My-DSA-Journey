```python
class Solution:
    def threeSumClosest(self, nums: List[int], target: int) -> int:
        nums.sort()
        max_diff=float('inf')
        res_sum=0

        for i in range(0,len(nums)-2):
            left=i+1
            right=len(nums)-1

            while left<right:
                total=nums[i]+nums[left]+nums[right]
                diff=abs(total-target)

                if diff<max_diff:
                    max_diff=diff
                    res_sum=total

                if total<target:
                    left=left+1

                elif total>target:
                    right=right-1

                else:
                    return total
        return res_sum
```
