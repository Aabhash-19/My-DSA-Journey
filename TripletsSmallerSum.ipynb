```python
class Solution:
    def countTriplets(self, sum, arr):
        arr.sort()
        ans=0
        
        for i in range(0,len(arr)-2):
            left=i+1
            right=len(arr)-1
            
            while left<right:
                total=arr[i]+arr[left]+arr[right]
                
                if total>=sum:
                    right=right-1
                
                else: 
                    ans=ans+(right-left)
                    left=left+1
        
        return ans
```
